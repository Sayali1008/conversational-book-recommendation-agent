```python
query = (
    "SELECT bg.book_id, COUNT(r.book_id) as rating_count "
    "FROM book_genres bg "
    "LEFT JOIN ratings r ON bg.book_id = r.book_id "
    "WHERE bg.genre_id = 23 "
    "GROUP BY bg.book_id "
    "ORDER BY rating_count DESC "
    "LIMIT 5"
)
pd.read_sql(query, conn)
```
